이 노트북에서 다양한 에이전트들을 테스트하고 개발할 수 있습니다.

```python
import asyncio
import sys
import os

# 프로젝트 루트를 Python path에 추가
sys.path.append('..')

from core.config import load_config, get_env_config
from core.logging import setup_logging, get_logger
from core.llm_client import LLMClient
from core.database import DatabaseClient
from agents.base.interfaces import AgentConfig, AgentResult
```

```python
# 로깅 설정
setup_logging(level="INFO")
logger = get_logger("notebook")

# 설정 로드
config = load_config()
logger.info("Configuration loaded successfully")
```

```python
# LLM 클라이언트 테스트
if config.llm.api_key:
    llm_client = LLMClient(config.llm)
    
    # 간단한 테스트
    response = await llm_client.generate_response("안녕하세요! 간단한 인사말을 해주세요.")
    print(f"LLM Response: {response}")
else:
    print("OpenAI API 키가 설정되지 않았습니다. LLM 테스트를 건너뜁니다.")
```

```python
# 데이터베이스 클라이언트 테스트 (설정이 있는 경우)
if config.database.username and config.database.password:
    db_client = DatabaseClient(config.database)
    
    try:
        tables = await db_client.get_all_tables()
        print(f"Available tables: {tables}")
    except Exception as e:
        print(f"Database connection failed: {e}")
else:
    print("데이터베이스 설정이 없습니다. DB 테스트를 건너뜁니다.")
```

```python
# 새로운 에이전트 클래스 테스트를 위한 템플릿
# 여기에 새로운 에이전트를 추가하고 테스트해보세요

class TestAgent:
    def __init__(self, name: str):
        self.name = name
        self.logger = get_logger(f"test_agent.{name}")
    
    async def test_function(self):
        self.logger.info(f"{self.name} agent is working!")
        return f"Hello from {self.name}!"

# 테스트 실행
test_agent = TestAgent("example")
result = await test_agent.test_function()
print(result)
```
